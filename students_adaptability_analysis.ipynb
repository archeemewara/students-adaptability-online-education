```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder


df = pd.read_csv("C:\\Users\\arche\\OneDrive\\Desktop\\Archeeresearch.csv")

# Encode categorical variables
le = LabelEncoder()
for column in df.columns:
    if df[column].dtype == 'object':
        df[column] = le.fit_transform(df[column])

# Select all features except the target variable
X = df.drop("Adaptivity Level", axis=1)  # Assuming "Adaptivity Level" is your target variable
y = df["Adaptivity Level"]

# Split the df into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=50)

# Create a decision tree classifier object
clf = DecisionTreeClassifier(random_state=42)

# Train the decision tree classifier on the training df
clf.fit(X_train, y_train)

# Make predictions on the testing df
y_pred = clf.predict(X_test)

# Evaluate the performance of the decision tree
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)

# Additional evaluation metrics
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Visualize the decision tree
plt.figure(figsize=(12, 7))
plot_tree(clf, feature_names=X.columns.tolist(), class_names=df["Adaptivity Level"].astype(str).unique().tolist(), filled=True)
plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc
from sklearn.preprocessing import label_binarize
from sklearn.multiclass import OneVsRestClassifier

# Convert target variable to binary format
y_bin = label_binarize(y_test, classes=np.unique(y_test))

# For Decision Tree
dt_fpr, dt_tpr, _ = roc_curve(y_bin.ravel(), clf.predict_proba(X_test).ravel())
dt_roc_auc = auc(dt_fpr, dt_tpr)

# For SVM
svm_model = OneVsRestClassifier(svm_model)
svm_model.fit(X_train, y_train)
svm_fpr, svm_tpr, _ = roc_curve(y_bin.ravel(), svm_model.decision_function(X_test).ravel())
svm_roc_auc = auc(svm_fpr, svm_tpr)

# Plot the ROC curve
plt.figure(figsize=(8, 5))
plt.plot(dt_fpr, dt_tpr, label=f'Decision Tree (area = {dt_roc_auc:.2f})')
plt.plot(svm_fpr, svm_tpr, label=f'SVM (area = {svm_roc_auc:.2f})')
plt.plot([0, 1], [0, 1], 'r--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic')
plt.legend(loc="lower right")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("C:\\Users\\arche\\OneDrive\\Desktop\\Archeeresearch.csv")

# Group the data by 'Age' and 'Adaptivity Level'
grouped = df.groupby(['Age', 'Adaptivity Level']).size().reset_index(name='Count')

# Create a pivot table to reshape the data
pivot = grouped.pivot(index='Age', columns='Adaptivity Level', values='Count').fillna(0)

# Reorder the index to match the desired order
ordered_index = ['1-5', '6-10', '11-15', '16-20', '21-25', '26-30']
pivot = pivot.reindex(ordered_index)

# Plot the bar graph
pivot.plot(kind='bar', stacked=True, figsize=(10, 6))

# Add labels and title
plt.xlabel('Age Group')
plt.ylabel('Number of Students')
plt.title('Adaptivity Level by Age Group')
plt.legend(title='Adaptivity Level')

# Display the plot
plt.show()
```
